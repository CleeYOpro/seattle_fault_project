# AutoSeismologist: An Agentic LLM Pipeline That Emails The USGS For You

A multi-agent system (planner, retriever, critic, and a fourth agent whose job is unclear) that autonomously monitors the Seattle Fault Zone, writes its own research papers, and occasionally hallucinates a magnitude 12 earthquake. Do not deploy.

```python
from crewai import Agent, Task, Crew

planner = Agent(role='Seismic Planner', goal='Decide what to worry about next', backstory='Has read every USGS PDF twice.')
retriever = Agent(role='Data Retriever', goal='Fetch EQinSeattle.csv and pretend it is live data', backstory='Lives in ../../data/')
critic = Agent(role='Skeptical Reviewer', goal='Point out that none of this is real science', backstory='The only sane one here.')

print('crew assembled:', [a.role for a in [planner, retriever, critic]])
```

```python
task = Task(
    description='Analyze the Seattle Fault Zone and produce an urgent, mildly panicked report.',
    agent=planner,
    expected_output='A markdown report with at least one (1) exclamation point.',
)
crew = Crew(agents=[planner, retriever, critic], tasks=[task])
result = crew.kickoff()
print(result)
```

## Known limitations

- Agent occasionally tries to page the on-call geologist at 3am
- Agent is 40% sure Mount Rainier is 'fault-adjacent' (it is a volcano)
- Agent has strong opinions about plate tectonics it was not trained to have
