# Deep Reinforcement Learning for Optimal Seattle Evacuation Routing

A PPO agent learns to route pedestrians away from unreinforced masonry buildings (URM) during a simulated M7.0 Seattle Fault rupture. Trained entirely in a gym environment that we definitely built and is definitely not just `random.randint` with extra steps.

```python
import gymnasium as gym
import numpy as np
from stable_baselines3 import PPO

class SeattleEvacuationEnv(gym.Env):
    def __init__(self):
        super().__init__()
        self.observation_space = gym.spaces.Box(low=-1, high=1, shape=(64,))
        self.action_space = gym.spaces.Discrete(8)  # 8 compass directions
        self.step_count = 0

    def reset(self, seed=None, options=None):
        self.step_count = 0
        return np.zeros(64, dtype=np.float32), {}

    def step(self, action):
        self.step_count += 1
        reward = 1.0 if action in (2, 3) else -0.1  # 'flee toward Lake Washington' bonus
        done = self.step_count >= 50
        return np.random.randn(64).astype(np.float32), reward, done, False, {}

env = SeattleEvacuationEnv()
model = PPO('MlpPolicy', env, verbose=0, n_steps=2048, batch_size=64, learning_rate=3e-4)
print('env + PPO agent ready')
```

```python
model.learn(total_timesteps=1_000_000)
print('training complete')
print('mean survival reward: 847.3 (units: entirely made up)')
print('agents successfully evacuated: 118% (yes, over 100%, do not ask)')
```
